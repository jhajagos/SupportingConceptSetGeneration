```python
def load_tables(sptr, metadata_dict, domains_to_load=["ohdsi", "concept", "prepared_source"]):
  sdf_dict = {}
  for domain in domains_to_load:
    print(f"Loading domain: {domain}")
    for table in metadata_dict[domain]:
      print(f"Loading table: {table}")
      sdf_dict[table] = sptr.read.parquet(metadata_dict[domain][table])
      sdf_dict[table].createOrReplaceTempView(table)
  return sdf_dict
```

### ICD10CM PT (Preferred terms; codable)

```python
umls_icd10_pt_sdf = spark.sql("""
with stys (
  select CUI, collect_list(STY) as STYs from MRSTY s group by CUI
)
select c.*, stys.STYs
  from MRCONSO c
    join stys on stys.CUI = c.CUI
where SAB='ICD10CM' and TTY in ('PT')""")
umls_icd10_pt_df = umls_icd10_pt_sdf.toPandas()
umls_icd10_pt_df["PREFIX_CODE"] = umls_icd10_pt_df["CODE"].str.slice(stop=3)
umls_icd10_pt_df
```

### Use header terms to build a hierarchy for selecting subset of terms

```python
umls_icd10_ht_sdf = spark.sql("""
select c.*
  from MRCONSO c
where SAB='ICD10CM' and TTY = 'HT' and CODE like '%-%'
and CODE != 'ICD-10-CM'
 order by CODE""")
umls_icd10_ht_df = umls_icd10_ht_sdf.toPandas()
umls_icd10_ht_df
```

```python
code_split_df = umls_icd10_ht_df["CODE"].str.split("-", expand=True)
code_split_df
start_letter = code_split_df[0].str.slice(stop=1)
start_number = code_split_df[0].str.slice(start=1)
end_letter = code_split_df[1].str.slice(stop=1)
end_number = code_split_df[1].str.slice(start=1)

umls_icd10_ht_parsed_df = umls_icd10_ht_df.copy()
umls_icd10_ht_parsed_df["start_range"] = code_split_df[0]
umls_icd10_ht_parsed_df["end_range"] = code_split_df[1]
umls_icd10_ht_parsed_df["start_letter"] = start_letter
umls_icd10_ht_parsed_df["start_number"] = start_number
umls_icd10_ht_parsed_df["end_letter"] = end_letter
umls_icd10_ht_parsed_df["end_number"] = end_number

umls_icd10_ht_parsed_df = umls_icd10_ht_parsed_df[["CUI", "CODE", "STR", "start_range", "start_letter", "start_number", "end_range", "end_letter", "end_number"]]
umls_icd10_ht_parsed_df
```

```python
letters = [chr(65 + i) for i in range(26)]
numbers = [("0" + str(i))[-2:]  for i in range(100)]
code_dict_list_expanded = []

for row_dict in umls_icd10_ht_parsed_df.to_dict(orient="records"):
  start_letter = row_dict["start_letter"]
  start_number = row_dict["start_number"]
  end_letter = row_dict["end_letter"]
  end_number = row_dict["end_number"]

  start_letter_i = letters.index(start_letter)
  end_letter_i = letters.index(end_letter)

  special_case = False
  codes_expanded = []
  try:
    int(start_number)
  except ValueError:
    special_case = True
    codes_expanded += [row_dict["start_range"]]

  if  not special_case:

    for i in range(start_letter_i, end_letter_i + 1):
      if start_letter == end_letter:
        try:
          int(end_number)
          codes_expanded += [f"{start_letter}{numbers[j]}" for j in range(int(start_number), int(end_number)+1)]

        except ValueError:
          codes_expanded += [f"{start_letter}{start_number}", f"{end_letter}{end_number}"]

      elif letters[i] == end_letter:
        codes_expanded += [f"{end_letter}{numbers[j]}" for j in range(0, int(end_number)+1)]
      elif letters[i] == start_letter:
        codes_expanded += [f"{letters[i]}{numbers[j]}" for j in range(int(start_number), 100)]
      else:
        codes_expanded += [f"{letters[i]}{numbers[j]}" for j in range(0, 100)]

  code_dict_list_expanded += [{"CUI": row_dict["CUI"], "CODE": row_dict["CODE"],
                               "STR": row_dict["STR"], "PREFIX_CODE": c} for c in codes_expanded]

codes_expanded_df = pd.DataFrame(code_dict_list_expanded)
codes_expanded_df
```

```python
joined_codes_expanded_df = codes_expanded_df.merge(umls_icd10_pt_df[["PREFIX_CODE"]], on="PREFIX_CODE").drop_duplicates()
joined_codes_expanded_df
```

```python
com_df_cols = list(com_df.columns)
```

```python

def generate_coding(df, columns):
  union_df = {}
  for c in columns:
    union_df[c] = df.copy()[df[c] == 1][["CODE","STR", "concept_id"]]
    union_df[c].columns = ["code", "description", "concept_id"]
    union_df[c]["code_type"] = "ICD10CM"
    union_df[c]["code_class"] = c

    union_df[c] = union_df[c][["code", "code_type", "description", "concept_id", "code_class"]]

  return pd.concat(union_df, axis=0)
```

```python
charlson_code_export_df = generate_coding(com_df, com_df_cols)
charlson_code_export_df
```

```python
elh_columns = list(elh_df.columns)
elih_columns = elh_columns[elh_columns.index("chf"):]
elih_columns
```

## Mapping ICD10CM to SNOMED

### Building maps

```python
snomed_sdf = spark.sql("""with c_snomed as (
  select * from concept c where c.vocabulary_id = 'SNOMED')
select * from c_snomed
""")
```

```python
concept_map_sdf = spark.sql("""with c_icd10cm as (
  select * from concept c where c.vocabulary_id = 'ICD10CM'),
    c_snomed as (
      select * from concept c where c.vocabulary_id = 'SNOMED')
  select distinct ci.concept_id, ci.concept_name, ci.domain_id, ci.concept_code, ci.domain_id, ci.concept_class_id,
    ci.vocabulary_id,
    cs.concept_id as mapped_concept_id, cs.concept_name as mapped_concept_name, cs.domain_id as mapped_domain_id,
    cs.concept_code as mapped_concept_code,
    cs.concept_class_id as mapped_concept_class_id, cs.vocabulary_id as mapped_vocabulary_id
  from c_icd10cm ci
  join concept_relationship cr on cr.concept_id_1 = ci.concept_id
  join c_snomed cs on cs.concept_id = cr.concept_id_2
  and cr.relationship_id = 'Maps to'
    """
)
concept_map_sdf.createOrReplaceTempView("concept_map")
concept_map_sdf.toPandas()
```

```python
concept_map_sdf.select("mapped_concept_id").drop_duplicates().toPandas()
```

```python
snomed_descendant_concept_map_sdf = spark.sql("""
select c1.concept_id, c1.concept_name, c1.vocabulary_id, c1.domain_id,
  ca.descendant_concept_id, c2.concept_name as descendant_concept_name,
  c2.concept_code as descendant_concept_code,
  c2.vocabulary_id as descendant_vocabulary_id, c2.domain_id as descendant_domain_id,
  min_levels_of_separation,
  max_levels_of_separation
from concept_ancestor ca
  join mapped_concept_ids mic on ca.ancestor_concept_id = mic.mapped_concept_id
  join concept c1 on c1.concept_id = ca.ancestor_concept_id
  join concept c2 on c2.concept_id = ca.descendant_concept_id
  where  c1.concept_id != c2.concept_id and c1.vocabulary_id = c2.vocabulary_id
  order by c1.concept_id, min_levels_of_separation, c2.concept_id

  """)
```
